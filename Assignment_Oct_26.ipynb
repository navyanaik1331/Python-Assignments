```python
'''
11. Leap Year Checker
Accept a year. First check whether it is divisible by 4. If yes, check whether it is divisible by 100.
If it is, check whether it is divisible by 400.
Display whether the year is a leap year.
'''
year = int(input("Enter a year: "))
if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("Leap year")
        else:
            print("Not a leap year")
    else:
        print("Leap year")
else:
    print("Not a leap year")
```

```python
'''
12. Triangle Classification
Accept three side lengths. First check whether they can form a valid triangle.
If valid, check whether all three sides are equal, exactly two sides are equal, or all sides are different.
'''
a = float(input("Enter the first side: "))
b = float(input("Enter the second side: "))
c = float(input("Enter the third side: "))
if a + b > c and a + c > b and b + c > a:
    if a == b == c:
        print("Equilateral Triangle")
    elif a == b or b == c or a == c:
        print("Isosceles Triangle")
    else:
        print("Scalene Triangle")
else:
    print("Invalid Triangle")
```

```python
'''
13. Employee Bonus Calculator
Accept years of experience and performance rating. If experience is at least 5 years, check the rating.
Give a 20% bonus for an excellent rating and 10% for a good rating.
If experience is below 5 years, give a 5% bonus only for an excellent rating.
'''
experience = int(input("Enter years of experience: "))
rating = input("Enter performance rating (Excellent/Good): ").lower()
if experience >= 5:
    if rating == "excellent":
        print("Bonus: 20%")
    elif rating == "good":
        print("Bonus: 10%")
    else:
        print("Invalid performance rating")
else:
    if rating == "excellent":
        print("Bonus: 5%")
    else:
        print("No bonus")
```

```python
'''
14. Password Strength Checker
Accept a password. If its length is at least 8 characters, check whether it contains a digit.
If it does, check whether it contains an uppercase letter.
Classify it as strong or moderate based on these conditions. Otherwise, classify it as weak.
'''
pwd = input('Enter a password')
uc = 0
lc = 0
sp = 0
nd = 0
if len(pwd)>=8 :
    for i in pwd:
        if i.isupper():
            uc+=1
        elif i.islower():
            lc+=1
        elif i.isdigit():
            nd+=1
        else:
            sp+=1
else :
    print('password length should be more than 7 characters')

if(uc>=1 and lc>=1 and nd>=1 and sp>=1):
    print('strong password')
else:
    print('Weak Password')
```

```python
'''
15. Hospital Consultation Priority
Accept a patient's temperature and oxygen saturation.
If oxygen saturation is below 90%, classify as high priority.
Otherwise, check whether the temperature is at least 39°C.
If so, classify as urgent; otherwise, classify as normal priority.
(Simplified programming exercise, not a medical triage tool.)
'''
temperature = float(input("Enter patient's temperature in Celsius: "))
oxygen = float(input("Enter oxygen saturation (%): "))

if oxygen < 90:
    print("High Priority")
else:
    if temperature >= 39:
        print("Urgent")
    else:
        print("Normal Priority")
```

```python
'''
16. Income Tax Slab Calculator
Accept annual taxable income.
If income is up to ₹3 lakh, tax is 0%.
Otherwise, check whether it is up to ₹7 lakh and apply a 5% rate to income above ₹3 lakh.
For income above ₹7 lakh, apply progressive slabs of 0%, 5%, and 10% to the respective portions.
Calculate the total tax.
'''

income = float(input("Enter annual taxable income: "))

if income <= 300000:
    tax = 0

elif income <= 700000:
    tax = (income - 300000) * 0.05

else:
    tax = (400000 * 0.05) + (income - 700000) * 0.10

print("Annual Taxable Income: ₹", income)
print("Total Income Tax: ₹", tax)
```

```python
'''
17. College Admission
Accept marks in Mathematics, Physics, and Chemistry.
If Mathematics is at least 60, check Physics. If Physics is at least 50, check Chemistry.
If Chemistry is at least 50 and the total is at least 180, display "Eligible"; otherwise, display "Not Eligible".
'''

maths = int(input("Enter Mathematics marks: "))
physics = int(input("Enter Physics marks: "))
chemistry = int(input("Enter Chemistry marks: "))

total = maths + physics + chemistry

if maths >= 60:
    if physics >= 50:
        if chemistry >= 50:
            if total >= 180:
                print("Eligible")
            else:
                print("Not Eligible")
        else:
            print("Not Eligible")
    else:
        print("Not Eligible")
else:
    print("Not Eligible")

print("Total Marks:", total)
```

```python
'''
18. Online Food Delivery
Accept order amount, distance, and whether the customer has a membership.
If the order amount is at least ₹500, check whether the distance is within 5 km.
Apply free delivery if within 5 km; otherwise, charge ₹40.
If the order is below ₹500, check membership status and calculate delivery charges accordingly.
'''

amount = float(input("Enter order amount: ₹"))
distance = float(input("Enter delivery distance in km: "))
membership = input("Are you a member? (yes/no): ").lower()

if amount >= 500:
    if distance <= 5:
        delivery_charge = 0
    else:
        delivery_charge = 40
else:
    if membership == "yes":
        delivery_charge = 20
    else:
        delivery_charge = 40

total_amount = amount + delivery_charge

print("Order Amount: ₹", amount)
print("Delivery Charge: ₹", delivery_charge)
print("Total Amount: ₹", total_amount)
```

```python
'''
20. Mini Banking System
Accept account type, account balance, and transaction type (deposit or withdrawal).
If the account type is savings, check the transaction type. For withdrawals, check minimum balance requirements;
for deposits, validate the deposit amount.
If the account type is current, apply a different minimum balance rule.
'''

account_type = input("Enter account type (savings/current): ").lower()
balance = float(input("Enter account balance: ₹"))
transaction = input("Enter transaction type (deposit/withdrawal): ").lower()
amount = float(input("Enter transaction amount: ₹"))

if account_type == "savings":
    if transaction == "deposit":
        if amount > 0:
            balance = balance + amount
            print("Deposit successful")
        else:
            print("Invalid deposit amount")

    elif transaction == "withdrawal":
        if balance - amount >= 1000:
            balance = balance - amount
            print("Withdrawal successful")
        else:
            print("Minimum balance of ₹1000 must be maintained")
    else:
        print("Invalid transaction type")

elif account_type == "current":
    if transaction == "deposit":
        if amount > 0:
            balance = balance + amount
            print("Deposit successful")
        else:
            print("Invalid deposit amount")

    elif transaction == "withdrawal":
        if balance - amount >= 5000:
            balance = balance - amount
            print("Withdrawal successful")
        else:
            print("Minimum balance of ₹5000 must be maintained")
    else:
        print("Invalid transaction type")

else:
    print("Invalid account type")

print("Current Account Balance: ₹", balance)
```
