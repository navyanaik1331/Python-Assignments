```python
n = 4
for i in range(1, n+1):
    if i % 2 == 1:
        val = 1
    else:
        val = 0
        
    j = 0
    while j < i:
        print(val, end=" ")
        if val == 1:
            val = 0
        else:
            val = 1
        j = j + 1
    print()
```

```python
n = 4
for i in range(1, n+1):
    s = n - i
    while s > 0:
        print(" ", end=" ")
        s = s - 1

    j = i
    while j >= 1:
        print(j, end=" ")
        j = j - 1

    k = 2
    while k <= i:
        print(k, end=" ")
        k = k + 1

    print()
```

```python
rows = 5
cols = 4

for i in range(rows):
    j = 0
    while (j < i) :
        print(" ",end="")
        j = j+1

    k = 0
    while (k < cols):
        print('*',end=' ')
        k = k+1
    print()
```

```python
n = 7
for i in range(n):
    j = 0
    while j < n:
        star = False

        if i == 0 or i == 6:
            if j == 0 or j == 6:
                star = True
        elif i == 1 or i == 5:
            if j == 1 or j == 5:
                star = True
        elif i == 2 or i == 4:
            if j == 0 or j == 2 or j == 4 or j == 6:
                star = True
        else:
            if j == 1 or j == 3 or j == 5:
                star = True

        if star:
            print('*',end=' ')
        else:
            print(' ',end=' ')

        j = j + 1
    print()
```
