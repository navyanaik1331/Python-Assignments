```python
# fibonnaci series using for loop
n = int(input("Enter the number of terms: ")) 
a = 0
b = 1
for i in range(n):
    print(a, end=" ")
    c = a + b
    a = b
    b = c
```

```python
# sum of a digit using for loop
n = int(input("Enter a number: "))
total = 0
for i in range(100):
    if n > 0:
        digit = n % 10
        total = total + digit
        n = n // 10

print("Sum of digits:", total)
```

```python
# check if given number is palindrom or not using for loop
n = int(input("Enter a number: "))
temp = n
reverse = 0
for i in range(100):
    if temp > 0:
        digit = temp % 10
        reverse = reverse * 10 + digit
        temp = temp // 10
if n >= 0 and n == reverse:
    print("Palindrome number")
else:
    print("Not a palindrome number")
```
