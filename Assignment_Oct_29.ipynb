```python
'''
1. Sum of Even and Odd Numbers Get a number n from the user and calculate the sum of all even numbers
and the sum of all odd numbers between 1 and n.
'''
n = int(input('Enter a number'))
esum = 0
osum = 0
for i in range(1,n+1):
    if i % 2 == 0:
        esum += i
    else:
        osum += i
print(f'Sum of all Even numbers is: {esum}')
print(f'Sum of all Odd numbers is: {osum}')
```

```python
'''
2. Count Digits in a Number Get an integer from the user and count the number of digits using a single for loop.
Do not convert the number into a string. Example: 58392 → 5 digits
'''
num = int(input('Enter an integer'))
count = 0
for i in range(num):
    if num > 0:
        count += 1
        num = num//10
print('Number of digits:',count)
```

```python
'''
3. Find the Second-Largest Number Given a list of distinct numbers, find the second-largest number
without using sort() or max().
'''
numbers = [10, 25, 8, 40, 30]
largest = numbers[0]
second_largest = numbers[1]
if second_largest > largest:
    largest, second_largest = second_largest, largest

for num in numbers[2:]:
    if num > largest:
        second_largest = largest
        largest = num

    elif num > second_largest:
        second_largest = num

print("Second-largest number:", second_largest)
```

```python
'''
4. Count Vowels and Consonants Get a string from the user and count the vowels and consonants separately.
Ignore spaces, digits, and special characters.
'''
text = input("Enter a string: ")
vowels = 0
consonants = 0
for ch in text.lower():
    if ch in "aeiou":
        vowels += 1
    elif ch.isalpha():
        consonants += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
```

```python
'''
5. Calculate Factorial Get a positive integer from the user and calculate its factorial using a single for loop.
Example: 5 → 120
'''
n = int(input('Enter an integer'))
fact = 1
for i in range(1, n+1):
    fact = fact*i
print(f'factorial of {n} is {fact}')
```

```python
'''
7. Calculate Electricity Bill Get the units consumed by each of five customers using a list.
Calculate each customer's bill using these rates:
* First 100 units: ₹2 per unit
* Next 100 units: ₹3 per unit
* Above 200 units: ₹5 per unit
'''
units_list = [80, 150, 250, 100, 300]
for i in range(5):
    units = units_list[i]
    if units <= 100:
        bill = units * 2

    elif units <= 200:
        bill = (100 * 2) + (units - 100) * 3

    else:
        bill = (100 * 2) + (100 * 3) + (units - 200) * 5

    print("Customer", i + 1, "Bill: ₹", bill)
```

```python
'''
8. Reverse a Number Get an integer from the user and reverse its digits using a single for loop.
Do not convert it into a string. Example: 12345 → 54321
'''
n = int(input("Enter a number: "))
temp = abs(n)
reverse = 0
for i in range(100):
    if temp > 0:
        digit = temp % 10
        reverse = reverse * 10 + digit
        temp = temp // 10
if n < 0:
    reverse = -reverse

print("Reversed number:", reverse)
```

```python
'''
9. Analyze Student Marks Given a list of marks, count the students who passed (50 or above), failed (below 50), and
scored distinction (75 or above).
'''
marks = [85, 45, 76, 32, 90, 60, 49, 75]
p = 0
f = 0
dist = 0
for mark in marks:
    if mark >= 50:
        p += 1
    else:
        f += 1

    if mark >= 75:
        dist += 1

print('Passed students:', p)
print('Failed students:', f)
print('Students with distinction:', dist)
```

```python
'''
10. Find Perfect Numbers Print all perfect numbers between 1 and 1000 using a single for loop.
A perfect number equals the sum of its proper divisors.Example: 6 is perfect because 1 + 2 + 3 = 6.
'''
for num in range(1, 1001):
    sum = 0
    for i in range(1, num):
        if num % i == 0:
            sum = sum + i
    if sum == num:
        print(num)
```
